### Business Case Scenario - Sample Superstore Dataset

```python
# Load Dataset
import pandas as pd
df = pd.read_csv(r"E:\Datasets\sample_superstore.csv",encoding='ISO-8859-1')
pd.set_option('display.max_columns', None)
df.head()
```

#### Qn1. What is the overall profit percentage across all sales?

```python
# Total Profit and Sales
total_profit = df['Profit'].sum()
total_sales = df['Sales'].sum()
```

```python
# Profit Percentage
profit_percent = (total_profit/total_sales)*100
print(f"Overall Profit Percentage: {profit_percent:.2f}%")
```

#### Qn 2. What category is contributing the most to profit?

```python
# Group by Category and sum profit
category_profit = df.groupby('Category')['Profit'].sum().sort_values(
                    ascending = False)
print("Profit by Category: ")
print(category_profit)
```

#### Qn 3: Which sub-category should be improved due to losses?

```python
# mine
sub_category_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values(
    ascending=True)
print(sub_category_profit)
```

```python
loss_making_subcategories = df.groupby('Sub-Category')['Profit'].sum()
loss_making = loss_making_subcategories[loss_making_subcategories<0].sort_values()

print("Sub Categories making a loss:")
print(loss_making)
```

#### Qn 4: What is the average shipping time?

```python
# Convert into date time
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
# taking their difference
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
# Averaging it
avg_shipping_days = df['Shipping Days'].mean()
print(f"Average Shipping Time: {avg_shipping_days:.2f} days")
```

#### Qn 5: In which region is the profit margin highest?

```python
region_data = df.groupby('Region')[['Profit', 'Sales']].sum()
region_data
```

```python
region_data['Profit%'] = (region_data['Profit'] / region_data['Sales'])*100
region_data
```

```python
region_data_sorted = region_data.sort_values(by='Profit%', ascending=False)
print("Profit Percentage by Region")
print(region_data_sorted[['Profit%']])
```

#### Qn 6: Which customer segment is most profitable?

```python
segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(
                    ascending=False)
print("Profit by Customer Segment: ")
print(segment_profit)
```
