### E-commerce Purchase Data

```python
# Load dataset
import pandas as pd
df = pd.read_csv(r"E:\Datasets\black_friday_sales.csv")
df.head()
```

#### Q1. What is the total revenue (sum of all purchases)?

```python
total_revenue = df['Purchase'].sum()
print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

#### Q2. What is the average purchase amount by gender?

```python
avg_by_gender = df.groupby('Gender')['Purchase'].mean()
print("Average Purchase by Gender: ")
print(avg_by_gender)
```

#### Q3. Which age group spends the most?

```python
avg_by_age = df.groupby('Age')['Purchase'].mean().sort_values(ascending=False)
print("Average Purchase by Age Group:")
print(avg_by_age)
```

### Q4. Which city category (A, B, C) shows highest total purchases?

```python
# highest total purchase --> total revenue --> sum
Purchase_by_city = df.groupby('City_Category')['Purchase'].sum().sort_values(
                    ascending=False)
print("Total Purchases by City Category: ")
print(Purchase_by_city)
```

#### Q5. What is the correlation between age group and purchase amount?

```python
print(df['Age'].unique())
```

```python
# Create ordinal mapping
age_map = {
    '0-17':0,
    '18-25':1,
    '26-35':2, 
    '36-45':3,
    '46-50':4, 
    '51-55':5,
    '55+':6
}
df['Age_Ordinal'] = df['Age'].map(age_map)  #  '0-17' into 0  as a new col
correlation = df[['Age_Ordinal', 'Purchase']].corr().iloc[0, 1]
print(f"Correlation between Age and Purchase: {correlation:.4f}")
```

#### Q6. What is the average purchase per product category?

```python
for col in ['Product_Category_1','Product_Category_2','Product_Category_3']:
    avg_purchase =df[df[col].notnull()].groupby(col)['Purchase'].mean().sort_values(
                    ascending=False)
    print(f"{avg_purchase.head()}\n")
```

#### Q7. Who purchases more - married or unmarried people?

```python
# more --> sum
Purchase_by_marraige = df.groupby('Marital_Status')['Purchase'].mean()
married_map = {0:'Unmarried',1:'Married'}

print("Average by Marital Status:")
for status, avg in Purchase_by_marraige.items():
    print(f"{married_map[status]}: {avg:.2f}")
```

#### Q8. Which Occupations spend the most?

```python
# occupation category spends the most --> avg
avg_by_occupation = df.groupby('Occupation')['Purchase'].mean().sort_values(
    ascending=False)
print("Average Purchase by Occupation: ")
print(avg_by_occupation.head())
```
